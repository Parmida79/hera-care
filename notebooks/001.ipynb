### These two datasets point to the same patient and in the same order

columns that exist in both datasets:
1. I   beta-HCG(mIU/mL) - identical
2. PCOS (Y/N) - identical
3. AMH(ng/mL) - identical (data in cell number 305 is string, all others are float)
4. II    beta-HCG(mIU/mL) - identical (data in cell number 123 have different values)

#### II    beta-HCG(mIU/mL)
* infertile - float
* fertile - object

#### AMH & I   beta-HCG(mIU/mL)
* both are object

#### PCOS (Y/N)
* both are int


### preferred approach: merge two datasets using 'Patient File No.'

```python
# Rename columns in infertile_df
infertile_df.rename(columns={
    'II    beta-HCG(mIU/mL)': 'Beta_HCG_II(mIU/mL)',
    '  I   beta-HCG(mIU/mL)': 'Beta_HCG_I(mIU/mL)',
}, inplace=True)

infertile_df.columns
```

```python
# Rename columns in infertile_df
fertile_df.rename(columns={
    'II    beta-HCG(mIU/mL)': 'Beta_HCG_II(mIU/mL)',
    '  I   beta-HCG(mIU/mL)': 'Beta_HCG_I(mIU/mL)',
}, inplace=True)

fertile_df.columns
```

```python
# Merge on 'Patient File No.' (use left join to keep all from fertile_df, add from infertile_df)
# Assume that fertile_df have newer data, thus the value in cell number 123 for column 'Beta_HCG_II(mIU/mL)' will be chosen
pcos_df = pd.merge(fertile_df, infertile_df[['Patient File No.']],
                       on='Patient File No.', how='left')

pcos_df.columns
```

```python
from sklearn.impute import SimpleImputer

# Use Imputer to replace null values - best practice for this dataset (numeric-heavy medical data)

# Medical data often has outliers (e.g., hormone levels), so median is safer than mean.
# numeric_cols = pcos_df.select_dtypes(include=['float64', 'int64']).columns
imputer = SimpleImputer(strategy='median')
pcos_df[pcos_df.columns] = imputer.fit_transform(pcos_df[pcos_df.columns])

# For categorical columns (e.g., 'Blood Group', if any), use mode
# categorical_cols = pcos_df.select_dtypes(include=['object']).columns
# if len(categorical_cols) > 0:
#     imputer_cat = SimpleImputer(strategy='most_frequent')
#     pcos_df[categorical_cols] = imputer_cat.fit_transform(pcos_df[categorical_cols])

# Verify no nulls left
pcos_df.isna().any()
```
